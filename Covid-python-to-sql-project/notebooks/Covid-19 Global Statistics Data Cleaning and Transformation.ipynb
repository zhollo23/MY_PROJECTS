```python
# Read the downloaded CSV file
import pandas as pd
df = pd.read_csv(r"C:\Users\zholl\Downloads\.kaggle\covid19_global_statistics_2026.csv")
#df.head(25)
```

```python
#Create new columns based on information you already have

active case rate
df['active_case_rate']=df['active_cases']/df['total_cases']

case fatlity rate
df['case_fatality_rate']=df['total_deaths']/df['total_cases']

cases per 100k people
df["cases_per_100k"] = (df["total_cases"] / df["population"]) * 100000

deaths per 100k people
df["deaths_per_100k"] = (df["total_deaths"] / df["population"]) * 100000

case growth rate
df["case_growth_rate"] = df["new_cases"] / df["total_cases"]
df = df.drop(columns = ["case_growth_rate"])

#How many continents are in this dataset? What are the continents?
df['continent'].unique()
df['continent'].nunique()
```

```python
#convert the date column data type (object) to a datetime data type
#df.dtypes
df['date'] = pd.to_datetime(df['date'],format='%Y-%m-%d')
```

```python
#load into SQL
#import sqlalchemy as sal
#engine = sal.create_engine('mssql://localhost/master?driver=')
#conn = engine.connect()

import urllib
import sqlalchemy as sal
from sqlalchemy import create_engine

params = urllib.parse.quote_plus(
    "Driver={ODBC Driver 17 for SQL Server};"
    "Server=localhost;"
    "Database=CovidDB;"
    "Trusted_Connection=yes;"
    "TrustServerCertificate=yes;"
)

engine = sal.create_engine(f"mssql+pyodbc:///?odbc_connect={params}")
```

```python
conn = engine.connect()
df.to_sql('Covid_nums',con=conn,index=False,if_exists='replace')
```
